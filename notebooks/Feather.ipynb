```python
import fireducks.pandas as pd
import sys
import os
```

```python
def read_feather_files_and_check_data(directory):
    """
    Lee todos los archivos .feather en el directorio especificado, los carga en un diccionario de DataFrames,
    cuenta cuántas filas contienen valores nulos y cuántas filas duplicadas hay en cada archivo.

    Args:
        directory (str): Ruta del directorio que contiene los archivos .feather.

    Returns:
        dict: Diccionario donde las claves son los nombres de los archivos (sin extensión)
              y los valores son tuplas con (DataFrame, filas con nulos, filas duplicadas).
    """
    directory = os.path.abspath(directory)
    
    if not os.path.exists(directory):
        print(f"El directorio {directory} no existe.")
        return {}
    
    # Filtrar solo archivos .feather en el directorio
    feather_files = [f for f in os.listdir(directory) if f.endswith('.feather')]
    if not feather_files:
        print("No se encontraron archivos .feather en el directorio.")
        return {}
    
    # Cargar cada archivo .feather en un DataFrame, contar nulos y duplicados
    dataframes = {}
    for file in feather_files:
        file_path = os.path.join(directory, file)
        file_name = os.path.splitext(file)[0]  # Nombre del archivo sin extensión
        
        try:
            # Cargar archivo Feather
            df = pd.read_feather(file_path)
            
            # Contar filas con nulos
            null_rows = df.isnull().any(axis=1).sum()
            
            # Contar filas duplicadas
            duplicate_rows = df.duplicated(subset=["timestamp", "fecha"]).sum()
            
            # Guardar DataFrame, filas con nulos y duplicados en el diccionario
            dataframes[file_name] = (df, null_rows, duplicate_rows)
            
            print(f"Archivo '{file}' cargado exitosamente con {len(df)} filas y {len(df.columns)} columnas.")
            print(f"Filas con valores nulos en '{file}': {null_rows}")
            print(f"Filas duplicadas en '{file}': {duplicate_rows}")
        except Exception as e:
            print(f"Error al leer el archivo '{file}': {str(e)}")
    
    return dataframes


# Uso del script
if __name__ == "__main__":
    directory_path = "../../primer_modelo"
    dataframes_with_checks = read_feather_files_and_check_data(directory_path)
    
    # Ejemplo: imprimir resumen final de archivos cargados, filas con nulos y duplicados
    print("\nResumen final:")
    for name, (df, null_rows, duplicate_rows) in dataframes_with_checks.items():
        print(f"Archivo '{name}': {len(df)} filas, {len(df.columns)} columnas, "
              f"{null_rows} filas con nulos, {duplicate_rows} filas duplicadas.")
```

```python
import os
import pandas as pd

def get_common_columns_in_feather_files(directory):
    """
    Encuentra las columnas comunes a todos los archivos .feather en el directorio especificado.

    Args:
        directory (str): Ruta del directorio que contiene los archivos .feather.

    Returns:
        set: Conjunto de columnas comunes a todos los archivos .feather.
        dict: Diccionario con las columnas de cada archivo.
    """
    directory = os.path.abspath(directory)
    
    if not os.path.exists(directory):
        print(f"El directorio {directory} no existe.")
        return set(), {}

    # Filtrar solo archivos .feather en el directorio
    feather_files = [f for f in os.listdir(directory) if f.endswith('.feather')]
    if not feather_files:
        print("No se encontraron archivos .feather en el directorio.")
        return set(), {}

    # Almacenar las columnas de cada archivo
    columns_per_file = {}
    common_columns = None

    for file in feather_files:
        file_path = os.path.join(directory, file)
        file_name = os.path.splitext(file)[0]  # Nombre del archivo sin extensión

        try:
            # Cargar archivo Feather
            df = pd.read_feather(file_path)
            file_columns = set(df.columns)
            columns_per_file[file_name] = file_columns

            # Actualizar el conjunto de columnas comunes
            if common_columns is None:
                common_columns = file_columns
            else:
                common_columns = common_columns.intersection(file_columns)

            print(f"Archivo '{file}' cargado exitosamente con {len(file_columns)} columnas.")
        except Exception as e:
            print(f"Error al leer el archivo '{file}': {str(e)}")
    
    return common_columns, columns_per_file


# Uso del script
if __name__ == "__main__":
    directory_path = "../../primer_modelo"
    common_columns, columns_per_file = get_common_columns_in_feather_files(directory_path)

    # Imprimir columnas comunes
    print("\n==================================")
    print("Columnas comunes a todos los datasets:")
    for col in sorted(common_columns):
        print(f"  - {col}")

    # Imprimir columnas por archivo
    print("\nColumnas por archivo:")
    for file_name, columns in columns_per_file.items():
        print(f"Archivo '{file_name}': {len(columns)} columnas")
        print(f"  {', '.join(sorted(columns))}")
```
